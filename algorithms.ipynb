# Top k most frequent elements

Time: O(n) to build the counter + O(m log m) for sorting, where m is unique elements. Worst case O(n log n)
Space: O(m) for the counter

```python
# O(n log n ) - time complexity
# O(n) - space complexity
def topKFrequent(self, nums: List[int], k: int) -> List[int]:
    counter = {}
    for num in nums:
        if num in counter:
            counter[num] += 1
        else:
            counter[num] = 1
    sorted_counter = sorted(list(counter.items()), key = lambda x: x[1], reverse=True)
    return [i[0] for i in sorted_counter[:k]]
```

Min-Heap approach — O(n log k) time:
Maintain a heap of size k with the k most frequent elements
More efficient when k is much smaller than the number of unique elements

```python
# O(n log k) - time complexity
# O(n) - space complexity
def topKFrequent(self, nums: List[int], k: int) -> List[int]:
    counter = {}
    for num in nums:
        if num in counter:
            counter[num] += 1
        else:
            counter[num] = 1
    res = []
    for key, val in counter.items():
        heapq.heappush(res, (val, key))
        if len(res) > k:
            heapq.heappop(res)
    return [i[1] for i in res]
```

Bucket Sort approach — O(n) time:
Create buckets where bucket[i] contains all elements with frequency i
Since max frequency is n, you only need n+1 buckets
This is the optimal solution

```python
# O(n) - time complexity
# O(n) - space complexity
def topKFrequent(self, nums: List[int], k: int) -> List[int]:
    counter = {}
    for num in nums:
        if num in counter:
            counter[num] += 1
        else:
            counter[num] = 1
    buckets = [[] for _ in range(len(nums) + 1)]
    for key in counter:
        buckets[counter[key]].append(key)
    res = []
    index = -1
    while len(res) < k:
        res.extend(buckets[index])
        index -= 1
    return res
```

```python
def findDuplicate(nums):
    slow, fast = 0, 0
    print(f"slow: index {slow}, value {nums[slow]}; fast: index {fast}, value {nums[fast]}")
    while True:
        slow = nums[slow]
        fast = nums[nums[fast]]
        print(f"slow: index {slow}, value {nums[slow]}; fast: index {fast}, value {nums[fast]}")
        if slow == fast:
            break

    slow = 0
    print(f"!slow: index {slow}, value {nums[slow]}; fast: index {fast}, value {nums[fast]}")
    while slow != fast:
        slow = nums[slow]
        fast = nums[fast]
        print(f"slow: index {slow}, value {nums[slow]}; fast: index {fast}, value {nums[fast]}")
    return slow
```

```python
# Task Summary
# Given a list of batches (where each batch is a list of dictionaries
# with "agent" and "task" keys) and a target key, return a set of all 
# distinct agents that share at least one batch with key, excluding key itself.

input = [
    [
        {"agent": "AgentA", "task": "t1"},
        {"agent": "AgentC", "task": "t3"},
        {"agent": "AgentA", "task": "t5"},
    ],
    [
        {"agent": "AgentB", "task": "t1"},
        {"agent": "AgentB", "task": "t3"},
        {"agent": "AgentA", "task": "t5"},
    ],
    [
        {"agent": "AgentD", "task": "t1"},
        {"agent": "AgentC", "task": "t3"},
        {"agent": "AgentA", "task": "t5"},
    ],
]

key = "AgentC"

result = {"AgentA", "AgentD"}

# def func(input, key):
#     return result
```

```python
def find_shared_agents(input, key):
    result = set()
    for batch in input:
        batch_agents = {entry["agent"] for entry in batch}
        if key in batch_agents:
            result = result.union(batch_agents)
    result = result - {key}
    return result
```

```python
# |a-b| <= d AND |b-c| <= d AND |a-c| <= d

#  Example
# 0 --------- c - e -------- a ----- b ----------------------------------------- 30 ->
#                    |_______ d ________|
#            |_______ d ______|

# Example
# d=10.1
# memory = []

# 2 -> None, memory = [2]
# 1 -> None, memory = [2,1]
# 100 -> None, memory = [2,1,100]
# 3 -> (1,2,3), memory = [100]
# 4 -> None, memory = [100, 4] 
# 2 -> None, memory = [100, 4, 2]
# 3 -> (2,3,4), memory = [100]

# {2}
# 6,7,11
# 0,0,1
# {0: [6,7], 1: [18,19], 2: [21,22]}

# [6,7,18,19,21,22]
```

```python
# setup for tests

def run_case(d, inputs, expected, solution_cls):
    sol = solution_cls(d)
    results = []
    for num in inputs:
        out = sol.find_triplets(num)
        results.append(tuple(sorted(out)) if out is not None else None)

    expected_norm = [tuple(sorted(e)) if e is not None else None for e in expected]
    ok = results == expected_norm
    print(f"{'PASS' if ok else 'FAIL'}: d={d} inputs={inputs}")
    if not ok:
        print(f"  expected: {expected_norm}")
        print(f"  got:      {results}")
    return ok


test_cases = [
    # from the walkthrough in the cell above
    dict(
        d=10.1,
        inputs=[2, 1, 100, 3, 4, 2, 3],
        expected=[None, None, None, (1, 2, 3), None, None, (2, 3, 4)],
    ),
    # exact-boundary distances (diff == d must still count)
    dict(
        d=5,
        inputs=[0, 5, 10, 4],
        expected=[None, None, None, (0, 4, 5)],
    ),
    # values stay too spread out for a triple to ever form
    dict(
        d=1,
        inputs=[0, 10, 20, 30],
        expected=[None, None, None, None],
    ),
    # negative numbers spanning zero
    dict(
        d=3,
        inputs=[-2, 0, 2, -1],
        expected=[None, None, None, (-2, -1, 0)],
    ),
    # floats / non-integer d
    dict(
        d=0.5,
        inputs=[1.0, 1.2, 1.4, 5.0],
        expected=[None, None, (1.0, 1.2, 1.4), None],
    ),
    # only the numbers left in memory after a match should be reusable later
    dict(
        d=10.1,
        inputs=[2, 1, 100, 3, 4, 2, 3, 100, 50],
        expected=[None, None, None, (1, 2, 3), None, None, (2, 3, 4), None, None],
    ),
]

def run_all(solution_cls):
    results = [run_case(solution_cls=solution_cls, **case) for case in test_cases]
    print(f"\n{sum(results)}/{len(results)} passed")
```

```python
# hint: floor(input/d)

from collections import defaultdict


class Solution():
    def __init__(self, d):
        self.holder = defaultdict(list)
        self.d = d
        
    def find_triplets(self, num): 
        bucket_key = num // self.d
        left, mid, right = self.holder[bucket_key-1], self.holder[bucket_key], self.holder[bucket_key+1]
        merged = left + mid + right + [num]
        merged.sort()
        l, r = 0, 2
        if len(merged) >= 3:
            while r < len(merged):
                if abs(merged[l]-merged[r]) <= self.d:
                    res = merged[l:r+1]
                    for neib in res:
                        if neib != num:
                            self.holder[neib//self.d].remove(neib)
                    return res
                l += 1
                r += 1
        self.holder[bucket_key].append(num) 
        return None
```

```python
run_all(Solution)
```

# Intersection of Three Sorted Arrays

Given three integer arrays `arr1`, `arr2` and `arr3` **sorted in strictly increasing order**, return a sorted array of only the integers that appeared in **all three** arrays.

Constraints:
- `1 <= arr1.length, arr2.length, arr3.length <= 1000`
- `1 <= arr1[i], arr2[i], arr3[i] <= 2000` (the tests below also try negatives and empties, so don't rely on this)

```python
# setup for tests

def run_case(arr1, arr2, arr3, expected, solution):
    got = solution(arr1, arr2, arr3)
    ok = got == expected
    print(f"{'PASS' if ok else 'FAIL'}: {arr1} {arr2} {arr3}")
    if not ok:
        print(f"  expected: {expected}")
        print(f"  got:      {got}")
    return ok


test_cases = [
    # example from the problem statement
    dict(arr1=[1, 2, 3, 4, 5], arr2=[1, 2, 5, 7, 9], arr3=[1, 3, 4, 5, 8], expected=[1, 5]),
    # large values, nothing in common
    dict(arr1=[197, 418, 523, 876, 1356], arr2=[501, 880, 1593, 1710, 1870], arr3=[521, 682, 1337, 1395, 1764], expected=[]),
    # no common elements at all
    dict(arr1=[1, 2, 3], arr2=[4, 5, 6], arr3=[7, 8, 9], expected=[]),
    # identical arrays
    dict(arr1=[1, 2, 3], arr2=[1, 2, 3], arr3=[1, 2, 3], expected=[1, 2, 3]),
    # element is in two arrays but missing from the third
    dict(arr1=[1, 2, 3], arr2=[2, 3, 4], arr3=[3, 4, 5], expected=[3]),
    # negative numbers
    dict(arr1=[-5, -3, 0, 2], arr2=[-3, 0, 4], arr3=[-4, -3, 0, 7], expected=[-3, 0]),
    # arrays of different lengths
    dict(arr1=[1], arr2=[1, 2, 3, 4, 5, 6], arr3=[0, 1], expected=[1]),
    # an empty array
    dict(arr1=[], arr2=[1, 2], arr3=[1, 2], expected=[]),
]

def run_all(solution):
    results = [run_case(solution=solution, **case) for case in test_cases]
    print(f"\n{sum(results)}/{len(results)} passed")
```

```python
def arraysIntersection(arr1, arr2, arr3):
    p1, p2, p3 = 0, 0, 0

    result = []
    while p1 < len(arr1) and p2 < len(arr2) and p3 < len(arr3):
        if arr1[p1] == arr2[p2] == arr3[p3]:
            result.append(arr1[p1])
            p1 += 1
            p2 += 1
            p3 += 1
            continue
        m = max(arr1[p1], arr2[p2], arr3[p3])
        if arr1[p1] < m:
            p1 += 1
        if arr2[p2] < m:
            p2 += 1
        if arr3[p3] < m:
            p3 += 1
    return result
    
```

```python
run_all(arraysIntersection)
```
